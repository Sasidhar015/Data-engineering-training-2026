```python
from pyspark.sql import functions as F
registrations_df = spark.table("event_registrations_source")
events_df = spark.table("event_master")
registrations_df.printSchema()
events_df.printSchema()
display(registrations_df)
display(events_df)
```

```python

selected_df = registrations_df.select("registration_id","attendee_name","tickets","unit_price")
display(selected_df)
```

```python
confirmed_large_df = registrations_df.filter(
    (F.col("status")=="Confirmed") & (F.col("tickets")>2)
)
display(confirmed_large_df)
```

```python
amount_df = registrations_df.withColumn("total_amount",F.col("tickets")*F.col("unit_price"))
display(amount_df)
```

```python
booking_category_df = amount_df.withColumn(
    "booking_category",
    F.when(F.col("total_amount")>=2000,"High").otherwise("Regular")
)
display(booking_category_df)
```

```python
joined_df = (
    registrations_df.filter(
        (F.col("status")=="Confirmed") & (F.col("tickets")>0) & (F.col("unit_price")>0)
    )
    .join(events_df,on="event_id",how="inner")
    .withColumn("revenue",F.col("tickets")*F.col("unit_price")))
city_revenue_df = joined_df.groupBy("city").agg(
    F.sum("revenue").alias("confirmed_ticket_revenue")
).orderBy("city")
display(city_revenue_df)
```

```python

pre_merge_version = spark.sql("DESCRIBE HISTORY event_registrations_work").first()["version"]
print("Pre-MERGE version:", pre_merge_version)
```

```python
%sql
MERGE INTO event_registrations_work AS t
USING incoming_registrations AS s
ON t.registration_id=s.registration_id
WHEN MATCHED THEN UPDATE SET
 t.event_id=s.event_id, t.attendee_name=s.attendee_name,
 t.registration_date=s.registration_date, t.ticket_type=s.ticket_type,
 t.tickets=s.tickets, t.unit_price=s.unit_price, t.status=s.status, t.channel=s.channel
WHEN NOT MATCHED THEN INSERT
(registration_id,event_id,attendee_name,registration_date,ticket_type,tickets,unit_price,status,channel)
VALUES
(s.registration_id,s.event_id,s.attendee_name,s.registration_date,s.ticket_type,
 s.tickets,s.unit_price,s.status,s.channel); 
```

```python
%sql
CREATE OR REPLACE VIEW event_revenue_view AS
SELECT m.event_name,m.city,
 SUM(CASE WHEN r.status='Confirmed' AND r.tickets>0 AND r.unit_price>0
          THEN r.tickets*r.unit_price ELSE 0 END) AS total_confirmed_revenue
FROM event_master m LEFT JOIN event_registrations_source r ON m.event_id=r.event_id
GROUP BY m.event_name,m.city; 
```

```python
%sql
CREATE OR REPLACE TEMP VIEW pending_registrations_view AS SELECT * FROM event_registrations_source WHERE status='Pending'; 
```

```python

global_view_supported = True
try:
    registrations_df.filter(F.col("ticket_type")=="VIP").createOrReplaceGlobalTempView("vip_registrations_view")
except Exception as e:
    global_view_supported = False
    print("Global temp view unavailable; using session temp view. Reason:", str(e)[:400])
    registrations_df.filter(F.col("ticket_type")=="VIP").createOrReplaceTempView("vip_registrations_view")
print("Global view supported:",global_view_supported)
```

```python

display(spark.table("event_revenue_view"))
display(spark.table("pending_registrations_view"))
if global_view_supported:
    display(spark.table("global_temp.vip_registrations_view"))
else:
    display(spark.table("vip_registrations_view"))
```
