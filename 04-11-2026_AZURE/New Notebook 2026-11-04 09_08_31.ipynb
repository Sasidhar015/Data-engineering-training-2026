```python
from pyspark.sql import functions as F
from pyspark.sql import Window
```

```python
csv_path = "/Volumes/<catalog>/<schema>/<volume>/cinema_bookings.csv"
json_path = "/Volumes/<catalog>/<schema>/<volume>/cinema_master.json"

target_table = "<catalog>.<schema>.cinema_bookings_cleaned"
```

```python
bookings_df = (spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv(csv_path))

```

```python
cinemas_df = spark.read.json(json_path)

display(bookings_df)
display(cinemas_df)
```

```python
bookings_df.printSchema()
cinemas_df.printSchema()
```

```python
display(bookings_df.limit(10))
print("Total booking rows:", bookings_df.count())
```

```python
display(bookings_df.select("booking_id", "movie_title", "tickets", "booking_status"))
```

```python
confirmed_df = bookings_df.filter(F.col("booking_status") == "Confirmed")
display(confirmed_df)
```

```python
display( bookings_df.filter(
        (F.col("tickets") > 3) & (F.col("price_per_ticket") > 200)))
```

```python
bookings_df.select(
    F.sum(F.col("discount_pct").isNull().cast("int")).alias("missing_discount_pct"),
    F.sum(F.col("payment_method").isNull().cast("int")).alias("missing_payment_method")).show()
```

```python
cleaned_df = bookings_df.fillna({
    "discount_pct": 0,
    "payment_method": "Unknown"})
display(cleaned_df)
```

```python
gross_df = cleaned_df.withColumn(
    "gross_amount",
    F.col("tickets") * F.col("price_per_ticket"))
display(gross_df)
```

```python
amount_df = gross_df.withColumn(
    "net_amount",
    F.round(
        F.col("gross_amount") * (1 - F.col("discount_pct") / 100.0),2))
display(amount_df)
```

```python
size_df = amount_df.withColumn(
    "booking_size",
    F.when(F.col("tickets") >= 4, "Large").otherwise("Regular"))
display(size_df)
```

```python
display(size_df.orderBy(F.col("net_amount").desc()).limit(5))
```

```python
display(
    size_df.groupBy("booking_status")
           .agg(F.count("*").alias("booking_count"))
           .orderBy("booking_status"))
```

```python
confirmed_movie_df = (
    size_df.filter(F.col("booking_status") == "Confirmed")
           .groupBy("movie_title")
           .agg(
               F.sum("tickets").alias("total_tickets"),
               F.round(F.sum("net_amount"), 2).alias("net_revenue")))
display(confirmed_movie_df)
```

```python
top_cinemas_df = (
    size_df.filter(F.col("booking_status") == "Confirmed")
           .groupBy("cinema_id")
           .agg(F.round(F.sum("net_amount"), 2).alias("confirmed_net_revenue"))
           .orderBy(F.col("confirmed_net_revenue").desc())
           .limit(3))
display(top_cinemas_df)
```

```python
joined_df = size_df.join(cinemas_df, on="cinema_id", how="left")
display(joined_df)
```

```python
display(
    joined_df.select(
        "cinema_name",
        F.col("location.city").alias("city"),
        "movie_title",
        "booking_status"))
```

```python
confirmed_city_revenue_df = (
    joined_df.filter(F.col("booking_status") == "Confirmed")
             .groupBy(F.col("location.city").alias("city"))
             .agg(F.round(F.sum("net_amount"), 2).alias("confirmed_net_revenue"))
             .orderBy(F.col("confirmed_net_revenue").desc()))
display(confirmed_city_revenue_df)

unmatched_cinemas_df = joined_df.filter(F.col("cinema_name").isNull())
display(unmatched_cinemas_df.select("booking_id", "cinema_id", "movie_title"))
```

```python
display(
    cinemas_df.select(
        "cinema_id",
        "cinema_name",
        F.col("location.zone").alias("zone")))
```

```python
facilities_df = cinemas_df.select(
    "cinema_id",
    "cinema_name",
    F.explode("facilities").alias("facility"))
display(facilities_df)
```

```python
dates_df = (
    size_df.withColumn("booking_date", F.to_date("booking_date"))
           .withColumn("show_date", F.to_date("show_date"))
           .withColumn(
               "days_between_booking_and_show",
               F.datediff(F.col("show_date"), F.col("booking_date"))))
display(dates_df)
```

```python
duplicate_ids_df = (
    bookings_df.groupBy("booking_id")
               .count()
               .filter(F.col("count") > 1))
display(duplicate_ids_df)

deduplicated_df = dates_df.dropDuplicates(["booking_id"])
print("Rows before deduplication:", dates_df.count())
print("Rows after deduplication:", deduplicated_df.count())
```

```python
movie_revenue_df = (
    deduplicated_df.filter(F.col("booking_status") == "Confirmed")
                   .groupBy("movie_title")
                   .agg(F.round(F.sum("net_amount"), 2).alias("confirmed_net_revenue")))
revenue_window = Window.orderBy(F.col("confirmed_net_revenue").desc())
ranked_movies_df = movie_revenue_df.withColumn(
    "revenue_rank",
    F.dense_rank().over(revenue_window))
display(ranked_movies_df.orderBy("revenue_rank", "movie_title"))
```

```python
final_df = deduplicated_df.join(cinemas_df, on="cinema_id", how="left")
 
table_name = f"{catalog}.{schema}.cinema_bookings_clean"
 
(
    final_df.write.format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(table_name)
)
```

```python
display(spark.sql(f"SELECT * FROM {table_name} LIMIT 5"))
```

**Final challenge**

```python
performance_report_df = (
    deduplicated_df
        .filter(F.col("booking_status") == "Confirmed")
        .join(cinemas_df, on="cinema_id", how="left")
        .groupBy(
            F.col("location.city").alias("city"),
            "movie_title"
        )
        .agg(
            F.countDistinct("booking_id").alias("confirmed_bookings"),
            F.sum("tickets").alias("total_tickets"),
            F.round(F.sum("net_amount"), 2).alias("confirmed_net_revenue")
        )
        .orderBy("city", F.col("confirmed_net_revenue").desc())
)
display(performance_report_df)
```
