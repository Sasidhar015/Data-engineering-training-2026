```python
%sql
CREATE OR REPLACE TABLE retail_pro(
    product_id INT,
    product_name STRING,
    price DECIMAL(10,2)
)
USING delta;
```

```python
%sql
insert into retail_pro values (1, 'product1', 100.00);
insert into retail_pro values (2, 'product2', 200.00);
insert into retail_pro values (3, 'product3', 300.00);
```

```python
%sql
CREATE OR REPLACE TEMP VIEW product_updates AS
SELECT *
FROM VALUES
    (2, 'Wireless Mouse', 950),
    (10, 'USB Hub', 1200)
AS updates(product_id, product_name, price);
```

```python
%sql
MERGE INTO retail_pro AS target
USING product_updates AS source
ON target.product_id = source.product_id

WHEN MATCHED THEN
    UPDATE SET
        target.product_name = source.product_name,
        target.price = source.price

WHEN NOT MATCHED THEN
    INSERT (product_id, product_name, price)
    VALUES (source.product_id, source.product_name, source.price);
        
```

```python
%sql
SELECT * FROM retail_pro
ORDER BY product_id;
```
