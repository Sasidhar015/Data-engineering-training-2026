**TYPE 1**

```python
%sql
CREATE OR REPLACE TABLE retail_pro(
    product_id INT,
    product_name STRING,
    price DECIMAL(10,2)
)
USING delta;
```

```python
%sql
insert into retail_pro values (1, 'product1', 100.00);
insert into retail_pro values (2, 'product2', 200.00);
insert into retail_pro values (3, 'product3', 300.00);
```

```python
%sql
select * from retail_pro
```

**TYPE 2 **

```python
data = [
    (4, "Monitor", 12000),
    (5, "Printer", 18000),
    (6, "Webcam", 2500)
]
 
df = spark.createDataFrame(
    data,
    ["product_id", "product_name", "price"]
)
 
df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("retail_products_pyspark")
 
display(spark.table("retail_products_pyspark"))
```

**TYPE 3**
Create Table AS Select CTAS
 

```python
%sql
CREATE OR REPLACE TABLE retail_products_ctas
USING DELTA
AS
SELECT *
FROM VALUES
    (7, 'Tablet', 25000),
    (8, 'Speaker', 5000),
    (9, 'Router', 3500)
AS products(product_id, product_name, price);
```

```python
%sql select * from retail_products_ctas
```
